                                          Condição dentro da função

```python
def verificar_aprovacao(nota):
    if nota >= 6:
        return "Aprovado"
    else:
        return "Reprovado"
nota = [7.5, 5.75, 6.0]
print(f"Nota {nota[0]}: {verificar_aprovacao(nota[0])}",  f"Nota {nota[1]}: {verificar_aprovacao(nota[1])}",  f"Nota {nota[2]}: {verificar_aprovacao(nota[2])}")
```

                                                  Função com lista

```python
def soma_lista(numeros):
    total = 0
    for numero in numeros:
        total = total + numero
    return total
numeros = [10, 20, 30, 40]
resultado_soma = soma_lista(numeros)

print(f"A lista de números é: {numeros}")
print(f"A soma dos elementos da lista é: {resultado_soma}")
```

                                                  Função com filtro

```python
def filtrar_maiores_idade(idades):
    maiores = []
    for idade in idades:
        if idade >= 18:
            maiores.append(idade)
    return maiores
idades = [15, 18, 20, 16, 25]
print(f"A lista de idades original é: {idades}")
print(f"As idades maiores ou iguais a 18 são: {filtrar_maiores_idade(idades)}")
```

                                        Função com parâmetro padrão

```python
def calcular_juros(valor, taxa=0.05):
  juros = valor * taxa
  return valor + juros
print(calcular_juros(1000))
print(calcular_juros(1000, 0.10))
```
