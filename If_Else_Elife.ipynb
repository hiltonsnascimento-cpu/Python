Crie um programa que peça a nota de um aluno.

O programa deve mostrar:

Nota maior ou igual a 7 - Aprovado

Nota entre 5 e 6.9 - Recuperação

Nota menor que 5 - Reprovado

Exemplo:

Digite sua nota: 6

Desafio: utilize if, elif e else.

```python
nota = float(input("Digite sua nota: "))
nota = float(f"{nota:.2f}")
if nota >= 7:
    print("Aprovado")
elif nota >= 5 and nota <= 6.99:
    print("Recuperação")
else:
    print("Reprovado")
```

Crie um programa que peça:

Nome

Idade

O programa deve permitir a entrada somente se a pessoa tiver 18 anos ou mais e estiver com
ingresso.

O programa deve perguntar:

Você possui ingresso? (sim/não)

Regras

Se:
idade >= 18 E
ingresso == "sim"
Mostrar: Entrada permitida!
Caso contrário: Entrada não permitida.
Dica: utilize if, else, input, int e and.

```python
nome = input("Digite seu nome: ")
idade = int(input("Digite sua idade: "))
ingresso = input("Você possui ingresso? (sim/não): ").lower()
if idade >= 18 and ingresso == "sim":
    print("Entrada permitida!")
else:
    print("Entrada não permitida.")
```

Crie um programa que peça:

Valor da compra

Se o cliente possui cartão da loja,  
o programa deverá dar 10% de desconto somente se:
A compra for maior ou igual a R$ 100
e o cliente possuir o cartão.

Pergunte:

- Digite o valor da compra:
- Possui cartão? (sim/não):

Dependendo da Resposta aplicar ou não os 10%.


```python
Valor_Compra = float(input("Digite o valor da compra: "))
possui_cartao = input("Possui cartão? (sim/não): ").lower()

if Valor_Compra >= 100 and possui_cartao == "sim":
    desconto = Valor_Compra * 0.10
    Valor_Total = Valor_Compra - desconto
    print("Você recebeu 10% de desconto!")
else:
    print("Você não possui desconto,por não possuir o cartão ou a compra não ser acima de R$100.00.")
    Valor_Total = Valor_Compra

print(f"Valor total a pagar: R$ {Valor_Total:.2f}")
```
