Gerenciamento de Projetos
Uma empresa de tecnologia precisa organizar os projetos em andamento.

```python
projetos = ["ERP Corporativo", "App Financeiro", "Portal do Cliente", "Sistema de Estoque", "Dashboard BI"]
print("Lista inicial de projetos:", projetos)
primeiro_projeto = projetos[0]
print("Primeiro projeto:", primeiro_projeto)
projetos.append("API de Integração")
print("Após adicionar 'API de Integração':", projetos)
projetos.remove("Sistema de Estoque")
print("Após remover 'Sistema de Estoque':", projetos)
projeto_removido_pop = projetos.pop(1)
print("Após remover projeto da posição 1:", projetos)
projetos[2]="Dashboard Analytics"
print("Após alterar projeto da posição 2:", projetos)
quantidade_projetos = len(projetos)
print("Quantidade total de projetos:", quantidade_projetos)
projetos.sort()
print("Projetos em ordem alfabética:", projetos)
projetos.reverse()
print("Projetos com ordem invertida:", projetos)
esta_na_lista = "ERP Corporativo" in projetos
print("ERP Corporativo' está na lista?:", esta_na_lista)
print("Lista final de projetos:", projetos)
```

Gerenciamento de Chamados

Uma empresa de tecnologia precisa organizar os chamados de suporte recebidos durante o dia.

```python
chamados = ["Erro no Sistema", "Acesso Bloqueado", "Falha na Rede", "Computador Lento", "Problema no E-mail"]
print("Lista inicial de chamados:", chamados)
primeiro_chamado = chamados[0]
print("Primeiro chamado:", primeiro_chamado)
chamados.append("Impressora Offline")
print('Chamado: "Impressora Offline" adicionado', chamados)
chamados.remove("Falha na Rede")
print('"Falha na Rede" Removido:', chamados)
chamado_removido = chamados.pop(1)
print("Chamado removido da posição 1:", chamados)
chamados[2] = "Computador sem Internet"
print("Número de Chamados:", len(chamados))
chamados.sort()
print("Chamados em ordem alfabética:", chamados)
chamados.reverse()
print("Chamados em ordem reversa:", chamados)
print("Erro no Sistema está na lista?: ", "Erro no Sistema" in chamados)
print("Lista final de chamados:", chamados)
```

Consulta de Clientes

Uma empresa possui uma lista de clientes cadastrados. Crie um programa que permita consultar se determinado cliente está cadastrado no sistema.

```python
clientes = ["Tech Solutions", "Nova Digital", "Alpha Sistemas", "Inova Corp", "Oscorp"]
nome_empresa = input("Digite o nome da empresa: ")
if nome_empresa in clientes:
    print(f"O cliente '{nome_empresa}' cadastrado!")
else:
    print(f"O cliente '{nome_empresa}' NÃO está cadastrado!")
```

Cadastro de Tecnologias
Uma empresa de tecnologia deseja cadastrar as tecnologias utilizadas em seus projetos.

```python
tecnologias = []
print("Por favor, digite 3 tecnologias para cadastrar:")
for i in range(3):
    tecnologia = input(f"Digite a tecnologia {i+1}: ")
    tecnologias.append(tecnologia)
print("\nLista de tecnologias cadastradas:", tecnologias)
tecnologia_para_consultar = input("\nDigite uma tecnologia para consultar: ")
if tecnologia_para_consultar in tecnologias:
    print("Tecnologia encontrada!")
else:
    print("Tecnologia não cadastrada!")
```
