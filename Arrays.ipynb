```python
nomes = ["Ana", "João", "Maria", "Carlos"]
print(nomes)
```

```python
nomes = ["Ana", "João", "Maria", "Carlos"]
print(nomes[-1])
```

```python
### Append
nomes = ["Ana", "Carlos"]
nomes.append("Joao")
print(nomes)
```

```python
###Remove
nomes = ["Ana", "João", "Maria", "Carlos"]
nomes.remove("Maria")
nomes.remove("João")
print(nomes)
```

```python
###pop
nomes = ["Ana", "João", "Maria", "Carlos"]
nomes.pop(-1)
print(nomes)
```

```python
###Lenght == len
nomes = ["Ana", "João", "Maria", "Carlos"]
print(len(nomes))
nomes.pop(-1)
print(len(nomes))
```

```python
###Sort
nomes = ["Ana", "João", "Maria", "Carlos"]
nomes.sort()
print(nomes)

numeros = [3,5,7,4,1,36,9,0,21]
numeros.sort()
print(numeros)
```

```python
###reverse
nomes = ["Ana", "João", "Maria", "Carlos"]
nomes.reverse()
print(nomes)
```

```python
###Alterando um valor
nomes = ["Ana", "João", "Maria", "Carlos"]
nomes[2] = "Joana"
print(nomes)

numeros = [3,5,7,4,1,36,9,0,21]
numeros[6]= 27
print(numeros)
```

```python
###Identificar ou verificar
nomes = ["Ana", "João", "Maria", "Carlos"]
print("Maria" in nomes)
print("Daniel" in nomes)
```

```python
nome = input("Digite seu nome: ").lower()
print(f"Seu nome é: {nome}")
nomes = ["Ana", "João", "Maria", "Carlos"]
if nome in nomes:
    print(f"Olá {nome}! Seu nome está na lista de nomes.")
else:
    print("Seu nome não está na lista de nomes.")
```
