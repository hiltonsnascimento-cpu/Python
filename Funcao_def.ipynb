```python
def calcula_media(notas):
    soma = 0
    for nota in notas:
        soma = soma+nota
    return soma / len(notas)
    notas_a = [7,8,6]
    notas_b = [10,9,8]
    print(f"{calcula_media(notas_a):.2f}")
    print(f"{calcula_media(notas_b):.2f}")
```

```python
### Calculadora dde desconto
def calcular(preco, desconto_percentual):
  valor_desconto = preco * (desconto_percentual/100)
  preco_final = preco - valor_desconto
  return preco_final
  produto = int(input("Digite o preço do produto: "))
  desconto = int(input("Digite o valor do desconto: "))
  preco_final = calcular(produto, desconto)
  print(f"O preço final do produto é R$ {preco_final:.2f}")
```

### Exemplo prático: Função para calcular preço com desconto
Vamos criar uma função que calcula o preço final de um produto aplicando uma taxa de desconto opcional (com valor padrão de 10%).

```python
def calcular_preco_final(preco_original, desconto=10):
    valor_desconto = preco_original * (desconto / 100)
    preco_final = preco_original - valor_desconto
    return preco_final

# Exemplo 1: Utilizando o desconto padrão (10%)
print(f"Preço final com desconto padrão: R$ {calcular_preco_final(100):.2f}")

# Exemplo 2: Definindo um desconto personalizado (25%)
print(f"Preço final com 25% de desconto: R$ {calcular_preco_final(100, 25):.2f}")
```
